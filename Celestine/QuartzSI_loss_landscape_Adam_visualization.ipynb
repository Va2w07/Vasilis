```python
# Simulate Reference pulse
def simulate_reference(L, deltat):
    
    toff=1.0e-11
    twidth=8.0e-13
    tdecay=1.0e-12
    scale=1.0e12
    x=np.zeros(L, float)
    for t in range(0,L):
        T=t*deltat - toff
        x[t]=-scale*T* np.exp(-(T/twidth)**2 - T/tdecay)

    return x
```

```python
def rts(n0, nj, Dj):
    """
    Calculate reflection and transfer coeffs for individual layers.
    returns tuple of r, t and t**2 - r**2 for layer j.
    """

    c = np.cos(nj*Dj)
    s = np.sin(nj*Dj)
    d = c + (0.5j) * (nj/n0 + n0/nj) * s
    r = (0.5j) * s * (n0/nj - nj/n0) / d
    t = 1.0/d
    # note transmission coeff phase is relative to n0 and kD
    # s phase which is used to compute reflection coeff is relative to input
    return (r, t*np.exp((1J)*n0*Dj), t*t-r*r)


def RTm(m, n0, layers):
    """
    Apply rts across all layers.
    Reflection and Transmission coefficients for m layers.
    """
    U=0.0; V=1.0; T=1.0; R=0.0

    for j in range(0,m):
        nj = layers[j][0]
        Dj = layers[j][1]
        (r,t,s) = rts(n0, nj, Dj)
        Vlast = V
        (U,V) = (r*V + s*U, V - r*U)
        T = T * t * Vlast / V
        R = U/V

    return (R,T)


def simulate_parallel(x, layers, deltat, noise_level):
    L = len(x)
# convolution done with 4*L points
    M = 4 * L
    N = 8 * L
# wave number for frequency point k is k*dk
    deltaf = 1.0/(N*deltat)
    dk = 2 * np.pi * deltaf / 2.9979e8
    m = len(layers)
    indices = np.array([l[0] for l in layers])
    thicknesses = np.array([l[1] for l in layers])
    T = np.zeros(N, complex)

    for k in range(0,M+1):
        kD = k*dk*thicknesses
        nkD = np.array([(indices[l],kD[l]) for l in range(0,m)])

        T[k] = RTm(m, 1.0, nkD)[1]

    for k in range(1,M):
        T[N-k]=np.conjugate(T[k])

    z = np.zeros(N, float)
    z[0:L] = x[0:L]
    X = np.fft.fft(z)/N
    Y = T*X
    y = N * np.fft.ifft(Y).real + noise_level * np.random.randn(N)
    
    return (T,y)
```

### With optimizer

```python
def log_cosh_loss(y_pred, y_test):
    # Log-Cosh(x) = log((exp(x) + exp(-x)) / 2)
    # A numerically stable version is:
    # log(cosh(x)) ≈ |x| + log(1 + exp(-2|x|)) - log(2)
    diff = y_pred - y_test
    return torch.mean(torch.log(torch.cosh(diff + 1e-12)))
```

```python
def find_optimal_params(initial_guess = (n_init, k_init, d_init), n_iters = 50000, print_step = 1000):

    n10, k10, d10, n20, k20, d20 = n_init[0], k_init[0], d_init[0], n_init[1], k_init[1], d_init[1]

    n1 = torch.tensor(1.0, requires_grad = True)
    k1 = torch.tensor(1.0, requires_grad = True)
    d1 = torch.tensor(1.0, requires_grad = True)
    n2 = torch.tensor(1.0, requires_grad = True)
    k2 = torch.tensor(1.0, requires_grad = True)
    d2 = torch.tensor(1.0, requires_grad = True)

#     k1 = torch.tensor(0.0, requires_grad = False)
#     k2 = torch.tensor(0.0, requires_grad = False)

    n1s, k1s, d1s  = [n1.clone().detach()*n10], [k1.clone().detach()*k10], [d1.clone().detach()*d10]
    n2s, k2s, d2s  = [n2.clone().detach()*n20], [k2.clone().detach()*k20], [d2.clone().detach()*d20]

    true_x_exp_tens = torch.tensor(true_x_exp, requires_grad=False)

    learning_rate = 1e-4
    optimizer = torch.optim.Adam([n1, k1, d1, n2, k2, d2], lr=learning_rate)

    layers = [(n1*n10 - 1j*k1*k10, d1*d10), (n2*n20 - 1j*k2*k20, d2*d20)]
    T_exp, y_exp = simulate_from_signal(x_exp, layers, deltat)

    loss = (true_x_exp_tens - y_exp[:L]).abs().mean() +  log_cosh_loss(true_x_exp_tens, y_exp[:L])
    
    losses = [loss.clone().detach()]
    
    print(f"Epoch={1:6d}, n1={n1.item()*n10:2e}, k1={k1.item()*k10:2e}, d1={d1.item()*d10:2e} - n2={n2.item()*n20:2e}, k2={k2.item()*k20:2e}, d2={d2.item()*d20:2e} - Loss={loss.item():2e}")
    
    for i_iter in range(1, n_iters):

#         if i_iter > 5000:
#             k1.requires_grad = True
#             k2.requires_grad = True
            
        optimizer.zero_grad()

        loss.backward()

        optimizer.step()
        
        layers = [(n1*n10 - 1j*k1*k10, d1*d10), (n2*n20 - 1j*k2*k20, d2*d20)]

        T_exp, y_exp = simulate_from_signal(x_exp, layers, deltat)

        loss = (true_x_exp_tens - y_exp[:L]).abs().mean() + log_cosh_loss(true_x_exp_tens, y_exp[:L])

        with torch.no_grad():

            losses.append(loss.clone().detach())

            if i_iter in [n_iters - 1] or (i_iter + 1)%print_step == 0:
                lr = optimizer.param_groups[0]["lr"]
                print(f"Epoch={i_iter+1:6d}, n1={n1.item()*n10:2e}, k1={k1.item()*k10:2e}, d1={d1.item()*d10:2e} - n2={n2.item()*n20:2e}, k2={k2.item()*k20:2e}, d2={d2.item()*d20:2e} - Loss={loss.item():2e}")

            n1s.append(n1.clone().detach()*n10)
            k1s.append(k1.clone().detach()*k10)
            d1s.append(d1.clone().detach()*d10)

            n2s.append(n2.clone().detach()*n20)
            k2s.append(k2.clone().detach()*k20)
            d2s.append(d2.clone().detach()*d20)

    # Plot history
    fig = plt.figure(figsize=(12, 4))
    plt.title("Optimization process")
    plt.plot(losses[1:])
    plt.ylabel('Loss')
    plt.xlabel('Iteration')
    plt.yscale('log')
    plt.show()

    return np.array([n1s, k1s, d1s, n2s, k2s, d2s, losses])
```

```python
def nkds_from_trajectory(trajectory):
    best_params = trajectory.T[-1][:-1].reshape(2,3).T
    labels = ['n', 'k', 'd']
    for i, vs in enumerate(best_params):
        print(f"{labels[i]}_init = [{', '.join([f'{v:6f}' for v in vs])}]")
    print()
    print(f"Loss: {trajectory.T[-1][-1]}")
    return best_params
```

```python
def calculate_3d_loss_landscape(x, y, x_iter = 100, y_iter = 100, pm = 0.5, original_params = None):
    loss_landscape = []

    if original_params is None:
        min_loss_traj = np.inf
        best_traj = 0
        for i, traj in enumerate(trajectories):
            if traj[-1][-1] < min_loss_traj:
                best_pars = traj.T[-1][:-1]
                best_traj = i
                min_loss_traj = traj[-1][-1]

        vs = {k:best_pars[v] for k,v in inds.items()}
    else:
        vs = original_params.copy()
        
    xs = np.linspace((1-pm)*vs[x], (1+pm)*vs[x], x_iter)
    ys = np.linspace((1-pm)*vs[y], (1+pm)*vs[y], y_iter)
    
    for xi in xs:
        for yi in ys:
            vs[x], vs[y] = xi, yi

            layers_try = [(vs['n1'] - 1j*vs['k1'], vs['d1']),
                          (vs['n2'] - 1j*vs['k2'], vs['d2'])]

            T_exp_try, y_exp_try = simulate_from_signal(x_exp, layers_try, deltat)
            y_exp_try = y_exp_try[:L]  #.detach().cpu().numpy()
            loss = log_cosh_loss(y_exp_try.detach(), true_x_exp)
#             loss = mae_loss(true_x_exp, y_exp_try)
            loss_landscape.append(loss)  

    X, Y = np.meshgrid(xs, ys)
    return X, Y, np.array(loss_landscape).reshape(x_iter, y_iter)
```

```python
def plot_3d_loss_landscape_and_trajectories(x, y, X, Y, loss_land, res = 200):

    fig = plt.figure(figsize=(6, 6))
    ax = fig.add_subplot(projection='3d')

    ax.plot_surface(X, Y, loss_land, cmap='terrain', alpha=0.6, antialiased=True, linewidth=0, rcount=res, ccount=res)

    for i, traj in enumerate(trajectories):
        ax.scatter(traj[inds[x]], traj[inds[y]], traj[-1], marker='.', color = f"C{i}", s = 0.1, label = f"Trajectory {i+1}")
        ax.scatter(traj[inds[x]][-1], traj[inds[y]][-1], traj[-1][-1], marker='*', color = f"C{i}")
    ax.set_xlabel(x)
    ax.set_ylabel(y)
    ax.set_zlabel('loss')
    plt.legend()
    plt.show()
```

## Better views of the landscape and the descent

The 3-D surfaces and 2×3 contour grid above have three limitations: (i) the trajectory points do not lie on the plotted
slice (the other four parameters differ at every point, so the dots float off the surface); (ii) 50 000 scatter points hide the
*direction* and *speed* of the descent; (iii) a linear colour scale flattens a loss that spans an order of magnitude.

The cells below use `loss_landscape_viz.py` (same folder) and work in **scaled coordinates**
$x_i = (p_i - p_i^{\rm opt})/p_i^{\rm nominal}$ — the coordinates Adam itself optimises here — so $n\sim2$, $\kappa\sim5\cdot10^{-3}$ and
$d\sim10^{-3}$ m can be compared on equal footing.

1. **Sensitivity geometry** — eigen-analysis of $J^\top J$ ($J=\partial y_{\rm sim}/\partial x$): which parameter combinations the data
   constrain, and the $n\,d$ degeneracy. (The loss is MAE-dominated, i.e. V-shaped, so a Hessian is ill-defined — a finite-difference
   Hessian of a cone even reports *zero* $n$–$d$ coupling.)
2. **Two-phase PCA landscapes** — Phase I (fast descent, per-start planes in curvature-normalised units) and Phase II (slow drift along the κ valley, shared plane).
   The path really lies in the plotted plane here, unlike the axis-aligned slices.
3. **Axis-aligned slices, fixed** — log colour, shared scale, per-parameter ranges, paths as lines, $n\,d=$ const overlay.
4. **Straight-line interpolation** start → optimum (barrier check) and **animation** of the descent.

```python
import loss_landscape_viz as lv
import importlib; importlib.reload(lv)

# --- wrap the notebook's simulator into the two callables the module needs -------------------
true_x_t = torch.tensor(true_x_exp, dtype=torch.float32)

def ysim_from_params(p):                      # p: torch tensor [6] = n1,k1,d1,n2,k2,d2  -> y_sim[:L]
    layers = [(p[0] - 1j*p[1], p[2]), (p[3] - 1j*p[4], p[5])]
    _, y = simulate_from_signal(x_exp, layers, deltat)
    return y[:L]

def loss_np(p):                               # p: array-like [6] -> float, same loss as find_optimal_params
    with torch.no_grad():
        y = ysim_from_params(torch.as_tensor(np.asarray(p, float), dtype=torch.float32))
        return ((true_x_t - y).abs().mean() + log_cosh_loss(true_x_t, y)).item()

# --- trajectories from the runs above -------------------------------------------------------------
trajs = [lv.from_notebook_trajectory(t, f'Start {i+1}') for i, t in enumerate(trajectories)]
ref, scale = lv.reference_and_scale(trajs)    # ref = best point found; scale = mean initial guess (nominal values)
print('optimum:', dict(zip(lv.PARAM_NAMES, np.round(ref, 6))))
print('loss   :', loss_np(ref))
```

### 1. Local geometry at the optimum — what the data can and cannot constrain

```python
G, J = lv.sensitivity_matrix(ysim_from_params, ref, scale)      # G = JᵀJ/L in scaled coords (autograd, ~1 s)
fig, (sv, V) = lv.plot_geometry(G, ref, scale, loss_fn=loss_np)
d = np.sqrt(np.diag(G)); np.set_printoptions(precision=3, suppress=True)
print('sensitivity correlation matrix (rows/cols n1 k1 d1 n2 k2 d2):'); print(G/np.outer(d, d))
print('σ1/σ6 =', sv[0]/sv[-1])
# fig.savefig('Quartzsi plots/geometry.png')
plt.show()
```

Reading the panels: σ₁ (all of $n_1,d_1,n_2,d_2$ together) is the total optical path — extremely well constrained by the pulse delay.
σ₃, σ₄ are the *anti-phase* $n$ vs $d$ combinations of each layer, ~10–100× weaker: this is the $n\,d$ degeneracy that is only broken by the
Fabry–Pérot echoes. σ₅, σ₆ are the two κ's: 3–4 orders of magnitude weaker than σ₁, which is why Adam spends 18 000 of its 20 000 iterations
drifting in κ. Panel (c) shows the loss is V-shaped (MAE), not parabolic — the side bumps along σ₁ at ±10 % are echo-order aliasing.

### 2. The descent in two phases (PCA planes the path actually lies in)

```python
metric = lv.curvature_metric(G)     # rescale axes so a unit step changes the pulse equally -> κ no longer dominates the PCA
fig, out = lv.plot_two_phase_landscape(loss_np, trajs, ref, scale, res=61, metric=metric)   # ~1 min for 4 planes at res=61
# fig.savefig('Quartzsi plots/two_phase_landscape.png')
plt.show()
b = out['boundary']; print('phase boundary at iteration', b)
```

Phase I planes are per start (the three early descents do not share a plane; each is faithful to its own loss within ~10–20 %).
The L-shaped paths are characteristic of Adam: every parameter moves at roughly the same normalised rate, so the path first follows
whichever parameter has the largest gradient, then turns. In Phase II all three starts have merged into the same shallow κ valley and
creep along it — the plane is faithful (see the diagnostics below), and the elongated, nearly flat contours show why 20 000 iterations
gain only 25 % in loss.

To check how faithful a plane is, compare the true loss along the path with the loss of its in-plane projection:

### Reusing this for other samples

Everything above only needs (a) a list of trajectories in the `{'params': [N,6], 'loss': [N], 'label': str}` format
(`lv.from_notebook_trajectory` converts the arrays returned by `find_optimal_params`), (b) `loss_np(p)` and (c) `ysim_from_params(p)`.
For a different number of layers, set `lv.PARAM_NAMES` / `lv.PARAM_TEX` accordingly and pass `pairs=` to `plot_pair_slices`.
